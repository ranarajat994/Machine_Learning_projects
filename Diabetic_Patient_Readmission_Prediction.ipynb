```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import recall_score, accuracy_score, classification_report, f1_score
from xgboost import XGBClassifier
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"D:\Data Analytics & Data Science\NetMax Technologies\DataSet From claude\diabetic_readmission_dataset.csv")
df.head(20)
```

```python
df.isnull().sum()
```

```python
# handling missing values

df["num_lab_procedures"] = df["num_lab_procedures"].fillna(df["num_lab_procedures"].median())

df["max_glu_serum"] =df["max_glu_serum"].fillna(df["max_glu_serum"].mode()[0])

df["A1Cresult"] = df["A1Cresult"].fillna(df["A1Cresult"].mode()[0])
```

```python
df.duplicated().sum()
```

```python
df.head()
```

```python
df["A1Cresult"].value_counts()
```

```python
# hanlding categorical data

label = LabelEncoder()

column = ["age","gender","race","admission_type", "primary_diagnosis","max_glu_serum","A1Cresult","insulin","change","diabetesMed"]
for item in column:
    df[item] = label.fit_transform(df[item])
```

```python
df.head()
```

```python
df.drop("patient_id", axis=1, inplace=True)
```

```python
df.head()
```

```python
# feature selection
x = df.drop(columns=["readmitted"])
y = df["readmitted"]
```

```python
from sklearn.model_selection import train_test_split


x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2, random_state=42)
```

```python
# using logistics regression

model_logistic_regressor = LogisticRegression(max_iter=100)

model_logistic_regressor.fit(x_train,y_train)
```

```python
y_pred_lr = model_logistic_regressor.predict(x_test)
acc_lr= accuracy_score(y_test,y_pred_lr)
print(f"Accuracy score with Logistic Regression: {acc_lr:.2f}")
```

```python
from sklearn.metrics import confusion_matrix


confusion_matrix(y_test,y_pred_lr)
```

```python
# scaling the data to improve the accuracy 

scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
model_logistic_regressor.fit(x_train_scaled,y_train)
```

```python
y_pred_scaled_lr = model_logistic_regressor.predict(x_test_scaled)
acc_scaled_lr = accuracy_score(y_test,y_pred_scaled_lr)
print(f"Accuracy score with scaled data (Logistic Regression) : {acc_scaled_lr:.2f}")
```

```python
confusion_matrix(y_test,y_pred_scaled_lr)
```

```python
# using xgboost ang grid search

import random

from sklearn.model_selection import GridSearchCV


best_para = {"max_depth" :[3,5,7,12],
             "n_estimators" : [200,250,300,350]}

model = RandomForestClassifier(random_state=42)

grid_search=GridSearchCV(estimator=model,
                         param_grid=best_para,
                         cv=5,
                         n_jobs=-1,
                         verbose=1)

grid_search.fit(x_train_scaled,y_train)
```

```python
best_model = grid_search.best_estimator_
```

```python
y_pred_xg = best_model.predict(x_test_scaled)
acc_xg = accuracy_score(y_test,y_pred_xg)
print(f"Accuracy score with XGBOOST: {acc_xg:.2f}")
```

```python
confusion_matrix(y_test,y_pred_xg)
```

```python
classification_report(y_test,y_pred_xg)
```
